**Block 1** clones the repo and installs the packages (about 1 minute). If Colab asks you to restart the session afterwards, do so (Runtime > Restart session) and re-run Block 1.

```python
import os, subprocess, sys

REPO = "admet-triage-panel"
if not os.path.exists("app.py"):
    if not os.path.exists(REPO):
        subprocess.run(["git", "clone", "-q", f"https://github.com/lai-wohjon/{REPO}.git"], check=True)
    os.chdir(REPO)
sys.path.insert(0, os.getcwd())
print("Using folder:", os.getcwd())

%pip install -q -r requirements.txt
```
